## **3. Сортировка данных**

10 треков с максимальным количеством прослушиваний

```python
top_10_tracks = df.nlargest(10, 'stream_count')
top_10_tracks
```

#### **Ответ на вопрос:** *какие **исполнители** и **жанры** встречаются среди самых прослушиваемых треков?*

*Среди самых прослушиваемых треков чаще всего встречаются **исполнители***

```python
artists_top = top_10_tracks['artist_name'].value_counts()
artists_top
```

*Среди самых прослушиваемых треков чаще всего встречаются **жанры***

```python
genres_top = top_10_tracks['genre'].value_counts()
genres_top
```

## **5. Основные статистические показатели**

Среднее и медиана

```python
mean_pop = df['popularity'].mean()
median_pop = df['popularity'].median()

print(f"Среднее значение popularity: {mean_pop:.2f}")
print(f"Медиана popularity: {median_pop:.2f}")
```

Минимум, максимум и стандартное отклонение

```python
min_pop = df['popularity'].min()
max_pop = df['popularity'].max()
std_pop = df['popularity'].std()

print(f" Минимум popularity: {min_pop}")
print(f" Максимум popularity: {max_pop}")
print(f" Стандартное отклонение popularity: {std_pop:.2f}")
```

## **6. Мода (наиболее часто встречающееся значение )**

Мода для genre

```python
genre_mode = df['genre'].mode()[0]
genre_counts = df['genre'].value_counts() #смотрим сколько раз жанр встречается
genre_counts.head(5)
```

Мода для release_year

```python
genre_mode = df['release_year'].mode()[0]
genre_counts = df['release_year'].value_counts().sort_index() #смотрим сколько раз жанр встречается
genre_counts.head(5)
```

## **7. Гистограмма**

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, edgecolor='black', alpha=0.7, color='steelblue')
plt.title('Распределение популярности треков (popularity)', fontsize=16, fontweight='bold')
plt.xlabel('Popularity', fontsize=12) #какое значение популярности имеет исполнитель
plt.ylabel('Частота', fontsize=12) #сколько исполнителей имеют такое значение популярности
plt.grid(True, alpha=0.3)
plt.show()
```
