```python
df = pd.read_csv('titanic.csv')
```

3. Первые 5 строк

```python
df.head(5)
```

4. Создание X (признаки)

```python
X = df[["Pclass", "Age", "Sex", "Fare", "SibSp", "Parch"]]
X.head()
```

5. Создание y (целевая переменная)

```python
y = df["Survived"]
y.head()
```

6. Проверка пропущенных значений

```python
X.isnull().sum()
```

 7. Обработка пропусков в Age

```python
X["Age"].median() # среднее значение. заменим прпоуски на него
```

```python
X["Age"] = X["Age"].fillna(X["Age"].median())
```

```python
X.isnull().sum()
```

8. Преобразование Sex в числовой вид

```python
X["Sex"] = X["Sex"].map({
    "male": 0,
    "female": 1
})
```

```python
X.head() # 0-мужчина, 1-женщина
```

 9. Разделение на train и test

```python
X_train, X_test, y_train, y_test = train_test_split( # 80% — обучение, 20% — тест
    X,
    y,
    test_size=0.2,
    random_state=42
)
```

10. Масштабирование признаков

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train) # fit + transform для train
```

```python
X_test_scaled = scaler.transform(X_test) # только transform для test
```

11. Вывод размеров

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```
