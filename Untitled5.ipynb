```python
df = pd.read_csv('titanic.csv')
```

Создание признаков

```python
Признаки = df[["Pclass", "Age", "Sex", "Fare", "SibSp", "Parch"]]
Признаки.head()
```

Создание целевой

```python
Целевая = df["Survived"]
Целевая.head()
```

Проверка пропущенных значений в признаках

```python
Признаки.isnull().sum()
```

Обработка пропусков в age

```python
Признаки["Age"].median() # среднее значение. заменим прпоуски на него
```

```python
Признаки["Age"] = Признаки["Age"].fillna(Признаки["Age"].median())
```

```python
Признаки.isnull().sum()
```

Преобразование sex в числовой вид

```python
Признаки["Sex"] = Признаки["Sex"].map({
    "male": 0,
    "female": 1
})
```

```python
Признаки.head() # 0-мужчина, 1-женщина
```

Формирование таблицы для корреляционного анализа

```python
data = pd.concat([Признаки, Целевая], axis=1)
data
```

Проверка полученной таблицы

```python
print("Столбцы в таблице:")
print(data.columns.tolist())
```

```python
print("Типы данных:")
print(data.dtypes)
```

```python
print("Типы данных:")
print(data.dtypes)
```

```python
print("Первые 5 строк:")
data.head()
```

Расчёт корреляционной матрицы

```python
corr_matrix = data.corr(numeric_only=True)
corr_matrix
```

Округленное значение

```python
corr_matrix_rounded = corr_matrix.round(2)
corr_matrix_rounded
```

Столбец Survived в коррелиции

```python
survived_corr = corr_matrix['Survived']
print(survived_corr)
```

Признак с наиболее сильной положительной корреляцией с Survived

```python
# Убираем корреляцию Survived с самой собой (она равна 1)
survived_corr_clean = corr_matrix['Survived'].drop('Survived')

# Наиболее сильная положительная корреляция = максимальное значение
max_pos_feature = survived_corr_clean.idxmax()
max_pos_value = survived_corr_clean.max()

print(f"Наиболее сильная положительная корреляция: "
      f"{max_pos_feature} = {max_pos_value:.2f}")
```

Признак с наиболее сильной отрицательной корреляцией с Survived

```python
# Наиболее сильная отрицательная корреляция = минимальное значение
max_neg_feature = survived_corr_clean.idxmin()
max_neg_value = survived_corr_clean.min()

print(f"Наиболее сильная отрицательная корреляция: "
      f"{max_neg_feature} = {max_neg_value:.2f}")
```

Построение тепловой карты

```python
# Размер графика
plt.figure(figsize=(10, 7))

# Тепловая карта
sns.heatmap(
    corr_matrix_rounded,       # данные — корреляционная матрица
    annot=True,                # вывод значений коэффициентов внутри ячеек
    cmap='coolwarm',           # цветовая палитра
    fmt='.2f',                 # формат чисел — 2 знака после запятой
    linewidths=0.5,            # линии между ячейками
    vmin=-1, vmax=1,           # диапазон значений от −1 до 1
    square=True,               # квадратные ячейки
    cbar_kws={'label': 'Коэффициент корреляции'}
)

plt.title('Тепловая карта корреляционной матрицы (Titanic)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Ответы на вопросы

1. Какую функцию библиотеки seaborn необходимо использовать?

sns.heatmap() - специальная функция для построения тепловых карт.

2. Какие данные передать в эту функцию?

Корреляционную матрицу corr_matrix_rounded (DataFrame с коэффициентами корреляции).

3. Какой параметр позволит вывести значения коэффициентов внутри ячеек?

annot=True (от англ. annotation - подпись).

4. Какой параметр отвечает за цветовую палитру?

cmap - например, 'coolwarm', 'RdBu_r', 'viridis', 'Blues'. Для корреляции удобно использовать дивергентные палитры (coolwarm, RdBu_r), так как значения могут быть отрицательными и положительными.

5. Как задать размер графика?

Двумя способами:

1) Через plt.figure(figsize=(ширина, высота)) до вызова heatmap.

2) Через параметр figsize в plt.subplots(figsize=...).
