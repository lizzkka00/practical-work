# **Задание №10**

```python
genre_sales_sorted = genre_sales.sort_values(ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(genre_sales_sorted.index, genre_sales_sorted.values)
plt.xticks(rotation=45, ha='right')
plt.xlabel("Жанр")
plt.ylabel("Мировые продажи, млн")
plt.title("Мировые продажи по жанрам")
plt.tight_layout()
plt.show()
```
